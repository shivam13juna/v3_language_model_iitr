### 3. Data preparation and cleaning

- The data is not always available in proper formats and also not with the required features to build a model.

Data cleaning is one of the important parts of machine learning.

Data cleaning is the process of preparing data for analysis by removing or modifying data that is incorrect, incomplete, irrelevant, duplicated, or improperly formatted.

1. Remove duplicate observations
2. Filter unwanted outliers
3. Fix structural errors
4. Fix missing data

- designation can be written in multiple way (ML engineer vs ML eng. vs machine learning engineer). (structural errors)

We have to deal with all kinds of scenarios.

* **Test set**
