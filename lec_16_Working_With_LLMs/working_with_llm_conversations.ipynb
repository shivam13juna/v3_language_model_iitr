```python
import os
from dotenv import load_dotenv

import textwrap



def pretty_print(*args):
    text = " ".join(str(arg) for arg in args)
    try:
        print(textwrap.fill(text, width=80))
    except Exception as e:
        print(text)  # fallback to normal print if text is not a string

        

load_dotenv('/Users/shivam13juna/Documents/scaler/iitr_classes/llm_ref/openai_key.env')  # reads .env file in the current directory

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError(
        "OPENAI_API_KEY not found! "
        "Make sure you have a .env file with: OPENAI_API_KEY=sk-..."
    )

pretty_print("API key loaded successfully.")
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)
pretty_print("OpenAI client ready.")
```

```python
models = client.models.list()

# These are the list of models I have access to, according to API key.
for m in models.data:
    pretty_print(m.id)
```

```python
response = client.chat.completions.create(
    model="gpt-5-nano",
    messages=[
        #{"role": "system", "content": "You are a very helpful assistant"},
        {"role": "user", "content": "Tell me about US vs IRAN War?"}
    ]
)

pretty_print(response.choices[0].message.content)
```

```python
current_messages = [
    {"role": "system", "content": "You are helpful efficient."},
    {"role": "user", "content": "What is color of apple?"}
]


response = client.chat.completions.create(
    model="gpt-5-nano",
    messages=current_messages
)

pretty_print(response.choices[0].message.content)
```

```python
# print current input tokens so far
import tiktoken
encoding = tiktoken.encoding_for_model("gpt-5-nano")
input_tokens = encoding.encode(" ".join([msg["content"] for msg in current_messages]))
pretty_print(f"Current input tokens: {len(input_tokens)}")
```

```python
current_messages.append({"role": "assistant", "content": response.choices[0].message.content})

current_messages.append({"role": "user", "content": "What did I just ask??"})

response = client.chat.completions.create(
    model="gpt-5-nano",
    messages=current_messages
)

pretty_print(response.choices[0].message.content)
```
