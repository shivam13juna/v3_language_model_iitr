
| Base | Size | Use it when... | Watch out |
|---|---|---|---|
| `python:3.12` | ~1 GB | you need a full toolchain | huge; rarely needed in final image |
| `python:3.12-slim` | ~150 MB | **default for AI apps** ✅ | may need a few `apt` libs |
| `python:3.12-alpine` | ~50 MB | pure-Python apps | ⚠️ **avoid for ML** (see below) |
| `nvidia/cuda:12.4.1-runtime-ubuntu22.04` | ~2–3 GB | GPU inference | only if you truly need CUDA |
